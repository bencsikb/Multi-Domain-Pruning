```python
import os
import glob
import pandas as pd
```

```python
reports_path =  "/data/blanka/REPORTS/YOLOv8x"

csv_files = sorted(glob.glob(os.path.join(reports_path, "*.csv")))
dfs = [pd.read_csv(f).assign(_source=os.path.basename(f)) for f in csv_files]

merged = pd.concat(dfs, ignore_index=True) if dfs else pd.DataFrame()
merged["dataset"] = merged["dataset"].astype(str).str.replace(r"\.yaml$", "", regex=True)

# nice table printing
pd.set_option("display.max_rows", 200)
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 180)
pd.set_option("display.expand_frame_repr", False)

print(merged.to_string(index=False))
```
