```python
#config_path = 'config/pruning/pruning_sampling.ini'
# data_path = "/data/blanka/DATASETS/SPN/YOLOv8x/data_shifted_zerofilled"
# label_path = "/data/blanka/DATASETS/SPN/YOLOv8x/label"

data_path = "/data/blanka/DATASETS/SPN/YOLOv8x_7030_modelwise_shifted_zerofilled_extra/training/data"
label_path = "/data/blanka/DATASETS/SPN/YOLOv8x_7030_modelwise_shifted_zerofilled_extra/training/label"

sample_handler = SampleHandler(data_path=data_path, label_path=label_path)
sample_handler.read_all_samples()
```

```python
all_samples = sample_handler.get_all_samples()
```

```python
import pandas as pd

dfs = [
    values[1]
        .reset_index()
        .assign(sample_key=key)
    for key, values in all_samples.items()
    if values and isinstance(values[1], pd.DataFrame)
]

labels_df = pd.concat(dfs, ignore_index=True)

print(labels_df)
```

```python
# df with last layers
final_labels_df = labels_df[labels_df['index'] == labels_df['index'].max()]
```

```python
final_labels_df.loc[:, "spars"] = 1 - final_labels_df["n_params"]/ final_labels_df["n_params_init"]
final_labels_df.loc[:, "dmap"] = 1 - final_labels_df["map50"]/ final_labels_df["map50_init"]
final_labels_df.loc[:, "spars_dmap_ratio"] = final_labels_df["spars"]/ final_labels_df["dmap"]
```

```python
final_labels_df.sort_values(by="spars_dmap_ratio", ascending=False, inplace=True)
pd.set_option("display.max_rows", None)
display(final_labels_df)   
```
